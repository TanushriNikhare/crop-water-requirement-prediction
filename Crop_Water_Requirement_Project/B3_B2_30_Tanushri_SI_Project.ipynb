```python
water = df_model["water_requirement_mm_day"]

print("Number of Observations:", len(water))
print("Mean Water Requirement:", water.mean())
print("Standard Deviation:", water.std())
```

```python
n = len(water)
sample_mean = water.mean()
sample_std = water.std()

standard_error = sample_std / np.sqrt(n)

print("Sample Size:", n)
print("Sample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
print("Standard Error:", standard_error)
```

```python
confidence_level = 0.95
alpha = 1 - confidence_level

z_critical = stats.norm.ppf(1 - alpha / 2)

margin_of_error = z_critical * standard_error

lower_limit = sample_mean - margin_of_error
upper_limit = sample_mean + margin_of_error

print("Z Critical Value:", z_critical)
print("Margin of Error:", margin_of_error)

print(
    f"95% Confidence Interval: "
    f"({lower_limit:.4f}, {upper_limit:.4f})"
)
```

```python
ci = stats.norm.interval(
    confidence=0.95,
    loc=sample_mean,
    scale=standard_error
)

print("95% Confidence Interval:", ci)
```

```python
plt.figure(figsize=(8, 5))

plt.errorbar(
    x=0,
    y=sample_mean,
    yerr=margin_of_error,
    fmt="o",
    capsize=10
)

plt.title("95% Confidence Interval for Mean Water Requirement")
plt.ylabel("Water Requirement (mm/day)")
plt.xticks([])

plt.show()
```

```python
hypothesized_mean = 5.0

z_statistic = (
    sample_mean - hypothesized_mean
) / (
    sample_std / np.sqrt(n)
)

p_value_z = 2 * (
    1 - stats.norm.cdf(abs(z_statistic))
)

print("Hypothesized Mean:", hypothesized_mean)
print("Z Statistic:", z_statistic)
print("P-value:", p_value_z)
```

```python
alpha = 0.05

if p_value_z < alpha:
    print("Reject the Null Hypothesis (H0)")
    print(
        "There is significant evidence that the "
        "population mean water requirement differs "
        "from 5.0 mm/day."
    )

else:
    print("Fail to Reject the Null Hypothesis (H0)")
    print(
        "There is insufficient evidence to conclude "
        "that the population mean water requirement "
        "differs from 5.0 mm/day."
    )
```

```python
t_statistic, p_value_t = stats.ttest_1samp(
    water,
    popmean=5.0
)

print("T Statistic:", t_statistic)
print("P-value:", p_value_t)
```

```python
alpha = 0.05

if p_value_t < alpha:
    print("Reject the Null Hypothesis (H0)")
    print(
        "There is statistically significant evidence "
        "that the population mean crop water requirement "
        "differs from 5.0 mm/day."
    )

else:
    print("Fail to Reject the Null Hypothesis (H0)")
    print(
        "There is insufficient statistical evidence "
        "to conclude that the population mean crop "
        "water requirement differs from 5.0 mm/day."
    )
```

```python
degrees_of_freedom = n - 1

t_critical = stats.t.ppf(
    0.975,
    df=degrees_of_freedom
)

margin_of_error_t = (
    t_critical * standard_error
)

lower_t = sample_mean - margin_of_error_t
upper_t = sample_mean + margin_of_error_t

print("Degrees of Freedom:", degrees_of_freedom)
print("T Critical Value:", t_critical)

print(
    f"95% t-Confidence Interval: "
    f"({lower_t:.4f}, {upper_t:.4f})"
)
```

```python
test_comparison = pd.DataFrame({
    "Test": [
        "Z-Test (Large-Sample Approximation)",
        "One-Sample t-Test"
    ],

    "Test Statistic": [
        z_statistic,
        t_statistic
    ],

    "P-Value": [
        p_value_z,
        p_value_t
    ],

    "Alpha": [
        0.05,
        0.05
    ]
})

test_comparison["Decision"] = np.where(
    test_comparison["P-Value"] < 0.05,
    "Reject H0",
    "Fail to Reject H0"
)

display(test_comparison)
```

```python
def one_sample_t_test(data, hypothesized_mean, alpha=0.05):

    t_stat, p_value = stats.ttest_1samp(
        data,
        popmean=hypothesized_mean
    )

    print("Hypothesized Mean:", hypothesized_mean)
    print("T Statistic:", t_stat)
    print("P-value:", p_value)

    if p_value < alpha:
        print("Decision: Reject H0")
    else:
        print("Decision: Fail to Reject H0")
```

```python
one_sample_t_test(
    df_model["water_requirement_mm_day"],
    5.0
)
```

```python
from scipy.stats import chi2_contingency
```

```python
contingency_table = pd.crosstab(
    df_model["crop_type"],
    df_model["soil_type"]
)

print("Contingency Table:")
display(contingency_table)
```

```python
chi2_stat, chi2_p, chi2_dof, expected = chi2_contingency(
    contingency_table
)

print("Chi-Square Statistic:", chi2_stat)
print("Degrees of Freedom:", chi2_dof)
print("P-value:", chi2_p)
```

```python
expected_table = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

print("Expected Frequencies:")
display(expected_table)
```

```python
alpha = 0.05

if chi2_p < alpha:
    print("Reject H0.")
    print(
        "There is statistically significant evidence "
        "of an association between crop type and soil type."
    )
else:
    print("Fail to Reject H0.")
    print(
        "There is insufficient evidence of an association "
        "between crop type and soil type."
    )
```
